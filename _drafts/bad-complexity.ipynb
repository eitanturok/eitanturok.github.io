Computer Scientists can't measure real-world difficulty

How do you measure how difficult something is? How hard is it to find the shortest path between two places (google maps), sort a list of numbers, or generate an image with AI?

Theoretical computer science gives us complexity theory – a framework to quantify how hard or complex a problem is. It’s very powerful. But also deeply flawed.

Is it easier to find love or watch a rom-com? (P vs NP in creative/abstract domains)

First off, it cannot measure the complexity of abstract concepts – like how hard is develop empathy, fall in love, or write a book. These things are generally not measurable and are rather subjective. Usually complexity theory only works for “algorithmic” problems, problems that have an objective solution and that we can make incremental, measurable progress towards them. For example, when sorting a list of numbers, we have an objectively correct notion of when numbers are sorted and we know we can measure how “sorted” these numbers, i.e. how many numbers are in the right spot. So this is still helpful for some things.

The classic classification system in P vs NP really is about the fact that it is harder to come with a solution than to check that that solution is correct. For example, it is hard to get a perfect score on a multiple-choice history test but very easy to check if your answers are correct (did you bubble A in for question 5?). How do we apply the distinction of solving a problem vs verifying the solution to these creative, abstract domains? For things like “how hard it is to fall in love” this means that if P = NP, you’d be able to determine if a couple is in love just as easily as you’d be able to fall in love yourself. It is weird applying to to these abstract domains. This hopefully makes the concept easier to understand. It is about doing the things VS checking the thing. It is easier to appreciate a good book than it is to write a good book. Applying P vs NP to these abstract creative domains makes it remarkably clear that P != NP. This just is an intuitive explanation for me.

“If P=NP, then the world would be a profoundly different place than we usually assume it to be. There would be no special value in “creative leaps,” no fundamental gap between solving a problem and recognizing the solution once it’s found. Everyone who could appreciate a symphony would be Mozart; everyone who could follow a step-by-step argument would be Gauss; everyone who could recognize a good investment strategy would be Warren Buffett.”

Can P = NP in mathematical domains but not in these creative domains? Since P is formally defined as “Problems in P represent the breadth of problems that are solvable in polynomial time by a deterministic Turing Machine” the question is do these creative problems satisfy the technical definition of a deterministic turning machine? I’m not sure.

Can you turn any creative problem into a mathetmical problem? Sure, but we don’t have the mathematical tools to do so, to properly formulate and quantity falling in love. You can measure dopamine, chemical processes, personal chemistry, toins of personality compatibility quizzes, the big five traits. And then you simply apply a matching algorithm. Algorithms like this exist, the stable marriage problem. Look at tinder’s algorithm. It might not be perfect but it is a way to quantify these things. But these are algorithms for matching people not actually determining if they are in love. 

Complexity cannot measure modern computers. Sequential vs Parallel.

Why don’t we measure how fast algorithms are in terms of seconds? Because the hardware changes. And so what is fast on your computer might be slow on my computer – not because the algorithm is inherently better but because your hardware is faster.

So when we measure how “fast” an algorithm is we measure its speed being proportional to the input size. This is hardware agnostic. IF you have n=100 objects, does it take time roughly proportional to 100, 100^2, or e^100? 

But this is stupid. It assumes everything is sequential,. Most computers are not purely sequential. They can do tings in parallel. (GPUs anyone?) Even normal MACs have multiple cores. And this notion of complexity breaks down when you can do so many things in parallel. Literally no computer today has one core. So hwy is the default way of measuring speed not taking into account parallelization?

Sequential vs parallel. In the make-believe world of computer science, we keep on pretending all computers are 100% sequential. The default measure of how complex algorithms are completes ignores the computational reality that even your calculator has multiple cores./ (JK I don’t think this is true.)

We need to change how we measure these things. Maybe this because time complexity is mainly sued among academics who have not touched a real computer since 1970s… I wish the parallel model of complexity would actually the default mode of computational complexity taught in every theoretical computer science class. It would make things slightly more practical.

I feel like being a twin embodies the sequential vs parallel dichotomy. There are two of you doing different things at the same time. This is like when your whole family cleans the kitchen in parallel it takes much shorter than if you were to do everything yourself, one step after another.

So why do theoreitians keep on measuring complexity in a sequential way? Are they lazy, don’t want to budge from their steadfast ways, or is it simply too complex to teach in an introductory computer science class

It is fundamentally the wrong paradigm.  Come on guys, enter the modern era… Just look at the innovations behind flash attention. They did complexity analysis for GPUs taking into account parallelization. And by actually performing analysis in the parallelizable regime, they got huge breakthroughs, enabling attention to be 5x faster [citation needed].

Why are we measuring the worst case?
"If you can’t handle me at my worst, then you sure as hell don’t deserve me at my best," - An algorithm telling a theoretical computer scientist to do worst-case analysis of algorithms

Traditional complexity theory measure algorithms with worst-case analysis. “Worst-case analysis summarizes the performance profile of an algorithm by it worst performance on any input of a given size”. This means if we are trying to sort a list of n=100 numbers, our algorithm might be really fast if the 100 numbers are all already sorted. But what if the numbers are in a random order? What happens if they are in a reverse order, smallest to largest instead of largest to smallest? Even though we are keeping the input size the same – we still have 100 numbers – we would measure the “performance” of this algorithm based on how fast it is on the worst possible arrangement of the input.

Why do we use worst-case analysis?

Robustness. Worst case analysis of algorithms implicit advocates for algorithms with the best-possible worst case performance. This ensures an algorithm is fast across any and all inputs, no matter how convoluted or adversarial it is. You know it will still adhere to this speed, guaranteed.
Worst case analysis is mathematically easier to prove than other kinds of analysis.

You may go through an entire undergraduate education in computer science without ever encountering the downside of worst-case analysis. But it certainly fails us in various ways.

Overly pessimistic: In linear programming, the simplex method has exponential time complexity [source] but is used far more often than the polynomial-time ellipsiod method. Analyzing the simplex method with worst-case analysis is overly pessimistic – there are a few contrived cases where the algorithm is slow but for all practical intents this algorithm is polynomial and in practice far more efficient than the ellpsiod method, which is embarrassingly slow. Worst-case complexity is misleading, like a stopwatch that measures how many steps you take as opposed to your actual time.

Ignores structured data. Clustering algorithms partition points into “coherent” groups. The k-median loss function is computaiotnlly intercatble, firmly residing in the NP-hard camp. Yet millions of clustering algorithms are run every day with remarkable success with incredibly simple procedures like Lloyd’s algorithm. How can we reconcile th eworst case inttability of clustering problems with the empirical success of relatibtly simple algorithms? The answer: clustering is hard only when it doesn’t matter. Sure, clustering will be hard for random, unstructured data points. But who cases? When clustering is used on real world data – points that represent images, documents, or proteins – meaniginful cluster likely exists and is to capture. Worst case analysis ignores structured data.

The problem. The unreasonable effectiveness of deep learning algorithms has thrown the gauntlet down to research in algorithm analysis and there is perhaps no other problem domain that calls out as loudly for a “beyond worst case” approach. Neural networks minimize a non convex loss function that are deemed computationally intractable by worst-case analysis. Yet stochastic gradient converges quickly to a local optimum and creates human-level intelligence. How can worst-case analysis lead to such incorrect conclusions?

We need beyond worst case analysis of algorithms. Worst-case algorithms grope the bottom of the barel. How do we fix this? Read Tim Rlughgardne’s BEyond the worst-case analysis of algorithms, one of my favorite.


Fine Grained Complexity

Complexity theory puts algorithms in two general camps:
P: the problems that are solvable in polynomial time by a deterministic Turing Machine
NP: the problems that are solvable within polynomial time by a non-deterministic Turing machine and whose solutions are verifiable by a deterministic Turing Machine in polynomial time
Let’s focus on the word polynomial. P are the problems  that are solvable quickly, in polynomial time, and NP, are the problems that must be solved in exponential time which modern computers cannot solve with an input of size greater n=100. But why are measuring algorithms with polynomial and exponential time? This distinction is huge!  This is helpful for large, sweeping statements and knowledge of the extremes. This is like having a stopwatch for a race that records the time every minute (polynomial) or every hour (exponential). But if I’m racing in the 100m dash, this clock is useless. I need a more fine-grained unit of measurement. The distinction between polynomial and exponential time is largely unhlepful for knowing if algorithms can be run on your computer. For that we need to know if algorithms are logarithmic, linear, or quadratic. This framework of polynomial vs exponential doens’t help as all of these dcateogires are within polynoimail time.

Williams pioneered the field of fine grained cok,explty to tackle this very problem [source] and has bore some algorithmically delicious fruit. The edit-distance problem underlies all of computational biology by determining how similar to string sof DNA are to each. Fine-grained complexity proved that this algorithm cannot be solved any faster than O(n^2). Standard complexity theory would simply tell us this algorithm runs in polynomial time O(n^c) for some c, and we wouldn’t know if this algorithm runs in O(n^100) or O(n^1.3) or something else, leaving it to the user to guess if this is feasible for execution on modern hardware.

Complexity theory is all about measuring how difficult or complex a problem is. But don’t you ever think we should keep it simple? After all, there is beauty in simplicity. In one of my favorite papers of all time,  Andrei Broder and Jorge Stolfi introduce an entirely new branch of computer science, simplicity theory. As they explain:

“For a long time theoretical Computer Science was concerned only with the analysis of either the worst case or the average case behavior of algorithms. This paper is the first to try to remedy the obvious discrimination against the study of the best case behavior…”


“In the study of reluctant algorithms, the performance of an algorithm A is better expressed by its inefficiency or best-case time, the minimum (as a function of n) over all inputs of size n of the running time of A. The simplexity of a problem is the maximum inefficiency among the reluctant algorithms that solve P. An algorithm is said to be pessimal for a problem P if the best-cast inefficiency of A is asymptotically equal to the simplexity of P.”


“One of the most important results of modern simplexity theory is the proof that the sorting problem can be solved in Ω(n log(n)/(2+)) time. This was the first problem to be shown to have non-polynomial simplexity. An elegant recursive algorithm that attains this inefficiency is the slowsort method below. The slowsort algorithm is a perfect illustration of the multiply and surrender paradigm, which is perhaps the single most important paradigm in the development of reluctant algorithms. The basic multiply and surrender strategy consists in replacing the problem at hand by two or more subproblems, each slightly simpler than the original, and continue multiplying subproblems and subsubproblems recursively in this fashion as long as possible. At some point the subproblems will all become so simple that their solution can no longer be postponed, and we will have to surrender. Experience shows that, in most cases, by the time this point is reached the total work will be substantially higher than what could have been wasted by a more direct approach.”

# Why I don't like complexity theory

How do you measure how difficult something is? 

Computer science tells us we can measure difficulty with complexity theory, a framework to quantify how hard or complex a problem is. It’s a very powerful framework and has guided researchers towards fast, efficient algorithms that have transformed the world over the course of decades.

However, I think that today complexity theory is a poor guide for finding new fast algorithms.

There are three main issues: complexity theory is overly pessimistic, ignores modern computer architecture, and provides blurry distinctions. 

These factors mean that complexity theory fails to accurately measure the performance of modern algorithms on modern computers. It is no longer a useful tool for algorithm designers. Complexity theory must adapt or die. My hope is that this blog encourages more folks to rethink how the standard complextiy theory class is taught in universities.

# Complexity theory is overly pessimistic.

> "If you can’t handle me at my worst, then you sure as hell don’t deserve me at my best" - Worst-case analysis of algorithms

In complexity theory, the default is to measure the speed of an algorithm by its *worst* possible performance on any set of inputs. It is often easier to prove results about the worst case performance than it is to prove things about the average or best case performance. Moreover, if we can prove that an algorithm is fast in the worst case, this is a pretty strong result, assuring us the algorithm will never ever be slow, no matter the input. So we use worst case analysis for it's ease of use and robustness.

However, many times worst case analysis is overly pessimistic. The worst case complexity may be entirely unrealistic, only occurring for some outrageously convoluted input that would never actually occur in the real world. The worst case complexity thus becomes an entirely meaningless indicator of actual performance. Indeed, sometimes you care about the average or best case performance such as when the experience of the typical user is paramount or when achieving the fastest possible performance is critical.

Let's go through several examples where worst case analysis is problematic.

The worst case time complexity for the quicksort algorithm is $O(n^2)$; imagine if you have to sort a deck of cards in ascending order but they are handed to you in descending order. On average, however, quicksort takes $O(n \log n)$ time to sort a uniformly random deck of cards. And in the best case, quicksort takes $O(n \log n)$ time, when the deck is handed to us already sorted in ascending order. So we can get vastly different perspectives on an algorithm by measuring its worst, average, or best case performance. Relying on the worst case analysis, we'd think quicksort is prohibitively expensive with a quadratic runtime. But the average runtime reassures us that for the average user, the quicksort algorithm is relatively fast and can perform pretty well.  That's why Python's sort function implements the quicksort algorithm. However, the best case case performance is $O(n \log n)$, not $O(n)$, telling us that if achieving the fastest possible runtime is important, one ought not to use quicksort.

In linear programming, the simplex method has exponential time complexity [source] but is used far more often than the polynomial-time ellipsiod method. Analyzing the simplex method with worst-case analysis is overly pessimistic – there are a few contrived cases where the algorithm is slow but for all practical intents this algorithm is polynomial and in practice far more efficient than the ellpsiod method, which is embarrassingly slow. Worst-case complexity is misleading, like a stopwatch that measures how many steps you take as opposed to your actual time.

The unreasonable effectiveness of deep learning algorithms has thrown the gauntlet down to research in algorithm analysis and there is perhaps no other problem domain that calls out as loudly for a “beyond worst case” approach. Neural networks minimize a nonconvex loss function that are deemed computationally intractable by worst-case analysis. Yet stochastic gradient converges quickly to a local optimum and creates human-level intelligence. How can worst-case analysis lead to such incorrect conclusions? Applying traditional worst case analysis to deep learning has by and large been a total failure. We need other methods.

All of these cases show the boundries and limitations of worst case analysis. Luckily, algorithm developers have responded in developing algorithms . Using only worst case analysis to measure difficulty is like only using a meter stick to measure distances; sometimes you need other measurements.

Tim Roughgarden's Beyond The Worst Case Analysis has done a great job of pushing the field to analyze algorithms beyond the worst case.

# Complexity theory ignores modern computer architecture.

Complexity theory arose in 60s when researchers modeled computers with a single processing units. However, every CPU since the 1970s has had multiple cores and can compute things in parallel. With today’s GPUs, there is more parallelization than ever. Yet standard complexity theory assumes our computers are stuck in the past where everything runs sequentially. This leads to many misleading results.

Here are some of examples.

One of the holy grails of complexity theory is a fast algorithm for multiplying matrices. The current record is held by Josh Alman, who I published a paper with, at O(n^omega) where omega=2.3…. People have been trying to drive this number down for decades and hundreds of papers have been published on this. But on nearly every modern computer, matrix multiplication is always implemented in parallel, each element of the resultant matrix is computed independently, taking $O(n/p)$ time where $p$ is the number of processors. Numpy even computes things in parallel. But complexity theory ignores this. It is driving innovation in the wrong area.

In the landmark paper Attention is All You Need, the training time of the transformer archectiure is reported as BLANK. RNNs, the preducrssor to transformers had a training time complexity of BLANK. Although these two surfaces it is BLANK for RNNs, the precursor to transformers. Yet transformers became the dominant archiecture because they are much faster than RNNs. the reason why transformers won out over RNNs is that they are inherently paralellizable during training. The time complexity of computing attention on $d$ computers is due to parallelization. Ignoring paralleization obscures the real reason why attention dominates today: massive parallelization.

# Complexity theory blurs distinctions

Complexity theory divides algorithms into two classes: "P" which is fast and "NP" which is slow. Algorithms in P run in polynomial time, e.g. $O(n^2)$, $O(4 n^{13})$ + $O(n^8 - 3.7 n^5 + 16 n)$. Algorithms in NP run in exponential time, e.g. $O(2^n)$, $O(5^{2n} + n)$, or $O(n^n)$. (This is a gross oversimplification of P vs NP but sufficies for now.)

Complexity theorists often don't care about the exact polynomial degree in P. They just care if the algorithm runtime is polynomial or exponential.
